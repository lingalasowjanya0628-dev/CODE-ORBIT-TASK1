```python
from google.colab import files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Upload Excel dataset
uploaded = files.upload()

# Get uploaded filename automatically
filename = list(uploaded.keys())[0]

# Read Excel file
df = pd.read_excel(filename)

print("Dataset loaded successfully!")
print("File:", filename)
print("Shape:", df.shape)

```

```python
print("\n========== DATASET INFORMATION ==========")

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()
```

```python
print("\n========== DATA QUALITY CHECK ==========")

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# Check unique values
print("\nUnique values:")
for column in df.columns:
    print("\n", column)
    print(df[column].unique()[:20])
```

```python
print("\n========== DATA TYPES AND VALUES ==========")

print("\nData types:")
print(df.dtypes)

print("\nQuantity statistics:")
print(df["Quantity"].describe())

print("\nTotal Sales statistics:")
print(df["Total_Sales"].describe())

print("\nOrder Date sample:")
print(df["Order_Date"].head(20))

print("\nPayment Methods:")
print(df["Payment_Method"].value_counts(dropna=False))
```

```python
print("\n========== NEGATIVE VALUE CHECK ==========")

# Negative Quantity
negative_quantity = df[df["Quantity"] < 0]

print("\nNegative Quantity values:")
display(negative_quantity[["Order_ID", "Quantity"]])

# Negative Total Sales
negative_sales = df[df["Total_Sales"] < 0]

print("\nNegative Total_Sales values:")
display(negative_sales[["Order_ID", "Total_Sales"]])
```

```python
print("\n========== REMOVE DUPLICATES ==========")

rows_before = len(df)

df = df.drop_duplicates()

rows_after = len(df)

print("Rows before removing duplicates:", rows_before)
print("Rows after removing duplicates:", rows_after)
print("Duplicate rows removed:", rows_before - rows_after)
```

```python
print("\n========== CLEAN TEXT DATA ==========")

text_columns = [
    "Customer_Name",
    "City",
    "Category",
    "Product",
    "Payment_Method"
]

# Remove extra spaces
for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# Standardize capitalization
for column in [
    "Customer_Name",
    "City",
    "Category",
    "Product"
]:
    df[column] = df[column].str.title()

# Standardize payment method
df["Payment_Method"] = df["Payment_Method"].str.title()

print("Text data cleaned successfully!")
```

```python
print("\n========== CONVERT DATA TYPES ==========")

# Convert Quantity to numeric
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

# Convert Total_Sales to numeric
df["Total_Sales"] = pd.to_numeric(
    df["Total_Sales"],
    errors="coerce"
)

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce",
    dayfirst=True
)

print("\nData types after conversion:")
print(df.dtypes)

```

```python
print("\n========== HANDLE MISSING VALUES ==========")

# Fill missing text values
for column in [
    "Customer_Name",
    "City",
    "Category",
    "Product",
    "Payment_Method"
]:
    df[column] = df[column].fillna("Unknown")

# Fill missing Email
df["Email"] = df["Email"].fillna("Unknown")

# Fill missing Quantity with median
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

# Fill missing Total Sales with median
df["Total_Sales"] = df["Total_Sales"].fillna(
    df["Total_Sales"].median()
)

# Fill missing Order Date with mode
if df["Order_Date"].isnull().sum() > 0:
    df["Order_Date"] = df["Order_Date"].fillna(
        df["Order_Date"].mode()[0]
    )

print("\nMissing values after cleaning:")
print(df.isnull().sum())

```

```python
print("\n========== HANDLE NEGATIVE VALUES ==========")

# Find median of valid Quantity values
quantity_median = df.loc[
    df["Quantity"] >= 0,
    "Quantity"
].median()

# Replace negative Quantity values
df.loc[
    df["Quantity"] < 0,
    "Quantity"
] = quantity_median


# Find median of valid Total Sales
sales_median = df.loc[
    df["Total_Sales"] >= 0,
    "Total_Sales"
].median()

# Replace negative Total Sales values
df.loc[
    df["Total_Sales"] < 0,
    "Total_Sales"
] = sales_median

print("Negative values handled successfully!")

print("\nNegative Quantity remaining:")
print((df["Quantity"] < 0).sum())

print("\nNegative Total_Sales remaining:")
print((df["Total_Sales"] < 0).sum())
```

```python
print("\n========== OUTLIER DETECTION ==========")

# Calculate Q1 and Q3
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)


# Find outliers
outliers = df[
    (df["Total_Sales"] < lower_bound) |
    (df["Total_Sales"] > upper_bound)
]

print("\nNumber of outliers:", len(outliers))

print("\nOutlier records:")
display(
    outliers[
        ["Order_ID", "Product", "Total_Sales"]
    ]
)

```

```python
print("\n========== OUTLIER VISUALIZATION ==========")

plt.figure(figsize=(10, 5))

plt.boxplot(df["Total_Sales"].dropna())

plt.title("Total Sales - Before Outlier Treatment")
plt.ylabel("Total Sales")

plt.show()
```

```python
print("\n========== HANDLE OUTLIERS ==========")

# Cap extreme values using IQR boundaries
df["Total_Sales"] = df["Total_Sales"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Outliers handled successfully!")
```

```python
print("\n========== FINAL VALIDATION ==========")

print("\nFinal dataset shape:")
print(df.shape)

print("\nFinal column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nNegative Quantity values:")
print((df["Quantity"] < 0).sum())

print("\nNegative Total_Sales values:")
print((df["Total_Sales"] < 0).sum())

print("\nFinal data types:")
print(df.dtypes)

```

```python
print("\n========== CLEANED DATASET ==========")

display(df.head(10))

```

```python
print("\n========== FINAL DATASET SUMMARY ==========")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")

for i, column in enumerate(df.columns, start=1):
    print(i, "-", column)
```
